```python
#1)Write a Python program using the map() function to calculate the square of each element in the following list:

numbers = [2, 4, 6, 8, 10]
squared_numbers = list(map(lambda x: x**2,numbers))
print(squared_numbers)
```

```python
#2)Write a Python program using the map() function to convert the following list of strings into integers:

num_string = ["10", "20", "30", "40", "50"]
int_list = list(map(int,num_string))

print(int_list)
```

```python
#3)Write a Python program using the filter() function to extract all even numbers from the following list:

numbers = [11, 24, 35, 42, 57, 68, 73, 80]
even_nos = list(filter(lambda x: x%2==0,numbers))

print(even_nos)
```

```python
#4) Write a Python program using the filter() function to extract all numbers greater than 50 from the following list:

numbers = [25, 65, 45, 80, 35, 95, 50]
result = list(filter(lambda x: x>50,numbers))

print(result)
```

```python
#5) Write a Python program using the reduce() function to calculate the sum of all elements in the following list:

from functools import reduce
a = [10, 20, 30, 40, 50]
sum = reduce(lambda x,y:x+y,a)

print(sum)
```

```python
#9.Create a 3-D NumPy array with shape (2, 3, 4) containing the numbers from 1 to 24. Write NumPy slicing statements to extract:

arr = np.arange(1,25).reshape(2,3,4)
print("Original array:",arr)

first_block = arr[0,:,:]
print(first_block)

print(arr[1, 0:2,:])

print(arr[:,:,2:])

```
