```python
#2)Given the array np.array([120, 150, 90, 200, 250, 180, 210]) , calculate mean,total sum

import numpy as np
arr = np.array([120, 150, 90, 200, 250, 180, 210])

total = 0

for i in arr:
  total = total + i
print(f"The total sum is:{total}")
print(f"The mean is: {np.mean(arr)}")
```

```python
#3) Write code to extract all numbers greater than 150

import numpy as np
arr = np.array([100,160,140,220,180]).tolist()

result = list(filter(lambda x: x>150,arr))
print(result)
```

```python
#4)Reshape a 1D array of 6 elements into a 2D array with 2 rows and 3 columns.
import numpy as np
arr= np.array([10,20,30,40,50,60])

reshaped_arr = arr.reshape(2,3)

print(f"Original array:{arr}")
print(f"Reshaped array:{reshaped_arr}")
```

```python
#5) Multiply all elements of np.array([10, 20, 30]) by 1.1 (a 10% increase).
import numpy as np
arr = np.array([10,20,30]).tolist()

result = list(map(lambda x: x*1.1,arr))
print(result)
```

```python
#matplotlib
#13) Write the basic Matplotlib code to plot days on the x-axis and sales on the y-axis.

import matplotlib.pyplot as plt

days = ['Mon','Tue','Wed']
sales = [100,150,200]

plt.plot(days,sales)
plt.title('Daily sales')
plt.xlabel('days')
plt.ylabel('sales')
plt.show()
```

```python
#17) Seaborn & Data visualization

import seaborn as sns
import pandas as pd

data = {
    "Day":["Mon","Tue","Wed"],
    "Sales":[100,150,200]
}
df = pd.DataFrame(data)

sns.set_theme(style="darkgrid")

sns.barplot(data=df, x='Day', y='Sales',width=0.3)
```

```python
# 19)Histogram

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

data = {
    "values":[10,20,30,45,50,60,70]
}
df = pd.DataFrame(data)

sns.histplot(data=df,x='values',kde=False)
plt.show()
```

```python
#20)Boxplot

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

data = {
    'Category': ['Electronics', 'Clothing', 'Electronics', 'Home', 'Clothing',
                 'Home', 'Electronics', 'Clothing', 'Home', 'Electronics'],
    'Sales': [250, 120, 300, 150, 90, 200, 450, 110, 180, 350]
}

df = pd.DataFrame(data)

custom_colors = ["#FF5733", "#33FF57", "#3357FF"]

sns.set_theme(style="whitegrid")

sns.boxplot(x="Category",y="Sales",data=df,hue="Category",palette=custom_colors,legend=True)
plt.title("Sales Distribution by category")
plt.show()
```
